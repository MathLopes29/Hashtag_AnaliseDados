```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
import numpy as np

print("Carregando e pré-processando o dataset IMDB...")

# 1. Parâmetros de pré-processamento
max_features = 10000  # Usar apenas as 10.000 palavras mais comuns
maxlen = 500          # Comprimento uniforme de 500 elementos para as sequências

# Carregar o dataset IMDB com o número especificado de palavras
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

print(f"Número de amostras de treinamento: {len(x_train)}")
print(f"Número de amostras de teste: {len(x_test)}")

# Padronizar as sequências para ter o mesmo comprimento
x_train = sequence.pad_sequences(x_train, maxlen=maxlen)
x_test = sequence.pad_sequences(x_test, maxlen=maxlen)

print(f"Shape dos dados de treinamento após padding: {x_train.shape}")
print(f"Shape dos dados de teste após padding: {x_test.shape}")

# 2. Definir o modelo da Rede Neural
print("\nDefinindo o modelo da Rede Neural...")
embedding_dim = 128 # Dimensão para a camada de Embedding

model = Sequential([
    Embedding(max_features, embedding_dim), # Camada de Embedding
    SimpleRNN(32), # Camada Recorrente Simples com 32 unidades
    Dense(1, activation='sigmoid') # Camada Densa de saída para classificação binária (sigmoid para 0 ou 1)
])

# 3. Compilar o modelo
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.summary()

# 4. Treinar o modelo
print("\nTreinando o modelo...")
epochs = 5
batch_size = 64 # Usar o batch_size solicitado

history = model.fit(x_train, y_train,
                    epochs=epochs,
                    batch_size=batch_size,
                    validation_data=(x_test, y_test)) # Usar dados de teste como validação

# 5. Testar o modelo e mostrar a performance
print("\nAvaliando o modelo nos dados de teste...")
loss, accuracy = model.evaluate(x_test, y_test)

print(f"\n### Performance do Modelo ###")
print(f"Acurácia do teste: {accuracy:.4f}")
print(f"Perda do teste: {loss:.4f}")
```
