### **Testes Estatísticos**

```python
# Teste de ADF
from statsmodels.tsa.stattools import adfuller

resultado_adf = adfuller(df["Price_Normalizado"].dropna())

print("Teste ADF")
print("Estatística:", resultado_adf[0])
print("P-valor:", resultado_adf[1])

if resultado_adf[1] < 0.05:
    print("Resultado: A série é ESTACIONÁRIA.")
else:
    print("Resultado: A série NÃO é ESTACIONÁRIA.")
```

```python
# Teste de KPSS
resultado_kpss = kpss(
    df["Price_Normalizado"].dropna(),
    regression="c",
    nlags="auto"
)

print("\nTeste KPSS")
print("Estatística:", resultado_kpss[0])
print("P-valor:", resultado_kpss[1])

if resultado_kpss[1] < 0.05:
    print("Resultado: A série NÃO é estacionária.\n")
else:
    print("Resultado: A série é ESTACIONÁRIA.\n")
```

##**Primeira Diferença**

```python
serie_diff = df_pib["PIB_Consumo_Normalizado"].diff().dropna()
print(serie_diff.head())
```

```python
resultado_adf = adfuller(serie_diff)

print("Teste ADF - Série Diferenciada")
print("Estatística:", resultado_adf[0])
print("P-valor:", resultado_adf[1])

if resultado_adf[1] < 0.05:
    print("A série diferenciada é ESTACIONÁRIA.")
else:
    print("A série diferenciada NÃO é estacionária.")
```

## **Segunda Diferença**

```python
resultado_adf_diff = adfuller(serie_diff)

if resultado_adf_diff[1] < 0.05:
    print("A série está estacionária!\n")
    serie_estacionaria = serie_diff
    numero_diferencas = 1

else:
    print("A série NÃO está estacionária após a primeira diferenciação!\n")
    serie_diff2 = serie_diff.diff().dropna()
    resultado_adf_diff2 = adfuller(serie_diff2)
    print("Aplicando segunda diferenciação...")

    if resultado_adf_diff2[1] < 0.05:
        print("A série está estacionária!")
        serie_estacionaria = serie_diff2
        numero_diferencas = 2
    else:
        print("A série ainda NÃO está estacionária!")
        serie_estacionaria = serie_diff2
        numero_diferencas = 2

print("\nNúmero de diferenciações:", numero_diferencas)
print("P-valor ADF:", adfuller(serie_estacionaria)[1])
```

```python
# ACF
plt.figure(figsize=(10, 5))
plot_acf(serie_estacionaria, lags=40, ax=plt.gca())
plt.title("ACF - PIB após Diferenciação")
plt.xlabel("Defasagem (Lag)")
plt.ylabel("Autocorrelação")
plt.show()

# PACF
plt.figure(figsize=(10, 5))
plot_pacf(serie_estacionaria, lags=40, ax=plt.gca(), method="ywm")
plt.title("PACF - PIB após Diferenciação")
plt.xlabel("Defasagem (Lag)")
plt.ylabel("Autocorrelação Parcial")
plt.show()
```

##**Conclusão**

Após a **aplicação da primeira diferenciação**, a série **não se tornou estacionária**, **conforme indicado pelo teste ADF**. Dessa forma, foi **aplicada uma segunda diferenciação**.

Após a segunda diferenciação, a série tornou-se **estacionária**, apresentando **p-valor inferior a 0,05**.

Nos gráficos de **ACF e PACF após a segunda diferenciação**, observa-se uma redução da dependência da série. A **ACF ainda apresenta alguns picos, principalmente nos lags múltiplos de 4**, indicando um possível padrão sazonal trimestral. Na **PACF**, os maiores picos aparecem nas primeiras defasagens, especialmente nos lags 1, 2 e 3.

Conclui-se que **foram necessárias duas diferenciações para tornar a série estacionária**. Apesar da estacionarização, os gráficos de **ACF e PACF indicam que a componente sazonal trimestral** ainda apresenta influência na série

```python
import pandas as pd
from bcb import sgs

df = sgs.get(
    {"Vendas_Varejo": 1455},
    start="2014-01-01",
    end="2020-01-31"
)


df = df.reset_index()
df = df.rename(columns={"Date": "data"})
df["periodo"] = df["data"].dt.to_period("M")

df = df[
    (df["periodo"] >= "2014-01") &
    (df["periodo"] <= "2020-01")
].copy()

df = df[["periodo", "Vendas_Varejo"]].reset_index(drop=True)
```

```python
df_copia = df.copy()
df_copia.rename(columns={"valor": "Vendas_Varejo"}, inplace=True)
df_copia.set_index("periodo", inplace=True)
df_copia.head()
```

#### **Conclusão**

ACF e PACF:
A análise da função de autocorrelação (ACF) evidencia **a presença de dependência temporal e de um padrão sazonal**. Observa-se uma autocorrelação positiva no primeiro lag, **indicando que os valores das vendas apresentam relação com os valores observados no mês anterior**. Destaca-se, entretanto, a **elevada autocorrelação no lag 12**, aproximadamente 0,75, que **ultrapassa o intervalo de confiança**. Como a série possui frequência mensal, o lag 12 representa **a relação entre um determinado mês e o mesmo mês do ano anterior**, constituindo uma forte evidência de sazonalidade anual. Também são observados picos **positivos nos lags 24 e 36**, correspondentes a **dois e três anos de defasagem**, respectivamente. **A presença de autocorrelações nos múltiplos de 12 reforça a existência de um padrão sazonal anual na série**, justificando a utilização de período sazonal igual a 12 na decomposição da série temporal.

A análise da função de autocorrelação parcial (PACF) **evidencia a existência de dependência temporal na série de vendas no varejo**. Observa-se um pico **positivo no primeiro lag**, **indicando dependência entre as vendas de um mês e as vendas do mês anterior**. Entretanto, o **resultado de maior destaque ocorre no lag 12**, que apresenta autocorrelação parcial próxima de 0,70 e ultrapassa significativamente o intervalo de confiança. Como a **série possui frequência mensal**, o lag 12 corresponde a um período de um ano, reforçando a evidência de um **componente sazonal anual identificada anteriormente pela ACF**. Também se observa um pico negativo no lag 13, que pode estar relacionado à estrutura da dependência sazonal da série. Em conjunto com a ACF, os resultados sugerem que a série apresenta forte componente sazonal com periodicidade de 12 meses.

A **ACF e a PACF indicam forte sazonalidade anual na série de vendas do varejo, evidenciada principalmente pelos elevados picos associados à defasagem de 12 meses**. Dessa forma, a periodicidade sazonal considerada nos modelos deve ser de 12 meses (s = 12).

##**Diferença Sazonal**

```python
from statsmodels.tsa.stattools import adfuller

df_copia["Vendas_Varejo"] = pd.to_numeric(
    df_copia["Vendas_Varejo"],
    errors="coerce"
)

# Diferença sazonal de 12 meses
serie_diff = df_copia["Vendas_Varejo"].diff(12).dropna()

print(serie_diff.head())
```

###**Testes Estatisticos**

```python
# Teste ADF
resultado_adf = adfuller(serie_diff)

print("\nTeste ADF - Série com Diferença Sazonal")
print("Estatística ADF:", resultado_adf[0])
print("P-valor:", resultado_adf[1])

# Decisão
if resultado_adf[1] < 0.05:
    print("A série diferenciada é ESTACIONÁRIA.")
else:
    print("A série diferenciada NÃO é estacionária.")
```

```python
# Teste de KPSS
resultado_kpss = kpss(
    serie_diff,
    regression="c",
    nlags="auto"
)

print("\nTeste KPSS")
print("Estatística:", resultado_kpss[0])
print("P-valor:", resultado_kpss[1])

if resultado_kpss[1] < 0.05:
    print("Resultado: A série NÃO é estacionária.\n")
else:
    print("Resultado: A série é ESTACIONÁRIA.\n")
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 5))
plot_acf(serie_diff, lags=40, ax=plt.gca())

plt.title("ACF - BACEN Vendas Varejo")
plt.xlabel("Defasagem (Lag)")
plt.ylabel("Autocorrelação")
plt.show()
```

```python
plt.figure(figsize=(20, 5))
plot_pacf(serie_diff, lags=30, ax=plt.gca(), method="ywm")

plt.title("PACF - BACEN Vendas Varejo")
plt.xlabel("Defasagem (Lag)")
plt.ylabel("Autocorrelação parcial")
plt.show()
```

###**Conclusão**
A diferença sazonal foi aplicada para remover a sazonalidade,**mas os seus testes mostram que ela não foi suficiente para tornar a série estacionária**.

##Segunda Tentativa

```python
serie_diff_final = serie_diff.diff().dropna()
```

```python
# ADF
resultado_adf = adfuller(serie_diff_final)

print("=== ADF ===")
print("Estatística:", resultado_adf[0])
print("P-valor:", resultado_adf[1])

if resultado_adf[1] < 0.05:
    print("A série diferenciada é ESTACIONÁRIA.")
else:
    print("A série diferenciada NÃO é estacionária.")


# KPSS
resultado_kpss = kpss(
    serie_diff_final,
    regression="c",
    nlags="auto"
)

print("\n=== KPSS ===")
print("Estatística:", resultado_kpss[0])
print("P-valor:", resultado_kpss[1])
if resultado_kpss[1] < 0.05:
    print("Resultado: A série NÃO é estacionária.\n")
else:
    print("Resultado: A série é ESTACIONÁRIA.\n")
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 5))
plot_acf(serie_diff_final, lags=40, ax=plt.gca())

plt.title("ACF - BACEN Vendas Varejo")
plt.xlabel("Defasagem (Lag)")
plt.ylabel("Autocorrelação")
plt.show()
```

```python
plt.figure(figsize=(20, 5))
plot_pacf(serie_diff_final, lags=30, ax=plt.gca(), method="ywm")

plt.title("PACF - BACEN Vendas Varejo")
plt.xlabel("Defasagem (Lag)")
plt.ylabel("Autocorrelação parcial")
plt.show()
```

###**Conclusão Final**

Após a aplicação da diferença sazonal e de uma diferença regular, o **teste ADF** rejeita a hipótese nula de raiz unitária, **indicando estacionariedade**. Entretanto, o **teste KPSS ainda rejeita a hipótese nula de estacionariedade ao nível de 5%**. Dessa forma, os testes apresentam resultados **divergentes, sendo necessário avaliar a série com cautela e considerar análises complementares antes de afirmar definitivamente que ela é estacionária**.
