# Séries Temporais - Introdução Prática

Nesta aula, vamos focar nos fundamentos práticos de Séries Temporais em Python usando a biblioteca Pandas. 

**O que é uma Série Temporal?**
Em essência, uma série temporal é um conjunto de dados ordenados no tempo. Em Python, a forma mais eficiente de trabalhar com esse tipo de dado é garantindo que o índice (index) do seu DataFrame ou Series seja um formato de data/hora (`datetime`).

### 1. Importando as Bibliotecas Essenciais

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 

# Configurando o tamanho padrão dos gráficos
from matplotlib.pylab import rcParams
rcParams['figure.figsize'] = 15, 6 
```

---
### 2. Criando uma Série Anual

Vamos simular dados que representam observações feitas uma vez por ano. Primeiro, geramos os valores.

```python
np.random.seed(102)
# Simulando 43 observações (ex: de 1980 a 2022)
dados_anuais = np.random.normal(0, 1, 43) 
```

O segredo para criar uma série temporal no Pandas é o comando `pd.date_range`. Nós informamos a data de início, a quantidade de períodos e a **frequência** (`freq`). 
Para dados anuais, usamos `freq='Y'` (Year end) ou `freq='AS'` (Year start).

```python
# Criando o índice de datas (Frequência Anual)
indice_anual = pd.date_range(start='1980', periods=len(dados_anuais), freq='YE')

# Combinando os dados e o índice em uma pd.Series
serie_anual = pd.Series(dados_anuais, index=indice_anual)
```

```python
indice_anual
```

```python
# Visualizando os primeiros registros e o tipo do índice
print(serie_anual.head())
print("\nTipo do Índice:", type(serie_anual.index))
```

```python
# Plotando a série
plt.plot(serie_anual)
plt.title('Série Temporal Simulada - Frequência Anual')
plt.show()
```

---
### 3. Criando uma Série Mensal

A lógica é exatamente a mesma, apenas alteramos a frequência para `'M'` (Month end) ou `'MS'` (Month start).

```python
np.random.seed(6)
# Simulando 96 meses (8 anos)
dados_mensais = np.random.normal(0, 1, 96)

# Criando o índice com frequência mensal, começando em Janeiro de 2015
indice_mensal = pd.date_range(start='2015-01-01', periods=len(dados_mensais), freq='ME')

serie_mensal = pd.Series(dados_mensais, index=indice_mensal)
```

```python
plt.plot(serie_mensal)
plt.title('Série Temporal Simulada - Frequência Mensal')
plt.show()
```

---
### 4. Criando uma Série Diária

Para dados diários, a frequência é `'D'` (Daily). Vamos simular um período mais curto (ex: 1 mês) para facilitar a visualização.

```python
np.random.seed(12)
# Simulando 31 dias
dados_diarios = np.random.normal(1, 2, 960)

# Criando o índice com frequência diária, começando em 1 de Janeiro de 2023
indice_diario = pd.date_range(start='2023-01-01', periods=len(dados_diarios), freq='D')

serie_diaria = pd.Series(dados_diarios, index=indice_diario)
```

```python
plt.plot(serie_diaria, marker='o') # Adicionando marcadores para ver os pontos diários
plt.title('Série Temporal Simulada - Frequência Diária (1 Mês)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

Carregando a série temporal airpassengers

```python
import seaborn as sns
```

```python
df = sns.load_dataset("flights")
df.head
```

Convertendo a coluna de data para o formato correto

```python
df['date'] = df['year'].astype(str) + '-' + df['month'].astype(str)
df['date'] = pd.to_datetime(df['date'])
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(x="date", y="passengers", data=df)
plt.xlabel('Data')
plt.ylabel('Número de Passageiros')
plt.title('Série Temporal Mensal de Passageiros de Avião')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
df.set_index('date', inplace=True)
```
