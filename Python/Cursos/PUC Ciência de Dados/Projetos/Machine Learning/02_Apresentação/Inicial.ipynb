# O que é Machine Learning?

Machine Learning (ML), ou Aprendizado de Máquina, é uma área da Inteligência Artificial que permite que computadores aprendam padrões a partir de dados, sem que todas as regras precisem ser programadas manualmente.

Imagine o seguinte problema:
"Você quer prever o preço de uma casa. Ao invés de escrever regras como:

* Se tiver 3 quartos → aumenta R$ 100.000
* Se tiver piscina → aumenta R$ 80.000
* Se estiver perto do centro → aumenta R$ 200.000

Você fornece milhares de exemplos para o computador"

| Área   | Quartos | Piscina | Preço   |
| ------ | ------- | ------- | ------- |
| 100 m² | 2       | Não     | 300.000 |
| 150 m² | 3       | Sim     | 550.000 |
| 80 m²  | 2       | Não     | 250.000 |


O algoritmo identifica sozinho os padrões existentes.

Esse processo é chamado de treinamento.
